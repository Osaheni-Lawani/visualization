```python
# load libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_csv("bikeshare_ridership_2026.csv")
```

```python
# Convert date column
df['Start_Time'] = pd.to_datetime(df['Start_Time'])
```

```python
# Create month variable
df['month'] = df['Start_Time'].dt.to_period('M')
```

```python
# Aggregate rides by month
monthly = df.groupby('month').size()
```

```python
# Plot
plt.figure(figsize=(10,5))
monthly.plot()
plt.title("Monthly Bike Share Ridership in Toronto")
plt.xlabel("Month")
plt.ylabel("Number of Rides")
plt.tight_layout()
plt.savefig("monthly_ridership.png")
plt.show()
```
