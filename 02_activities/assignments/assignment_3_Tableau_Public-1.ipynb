    > Who is your intended audience? 
    
    The intended audience includes city planners, transportation policymakers, Bike Share Toronto administrators, researchers, and members of the public interested in sustainable transportation. These stakeholders may use the visualization to understand how bike-sharing services are utilized across the city and identify patterns that can inform operational and policy decisions.

    > What information or message are you trying to convey with your visualization? 
    
    The primary objective of this visualization is to communicate ridership patterns among Bike Share Toronto users, with a particular focus on differences between annual members and casual users. The dashboard highlights several key trends:
    
    1. Annual members consistently account for a larger proportion of trips than casual riders.
    
    2. Ridership increases during warmer months, peaking during the summer (June to August), suggesting a strong seasonal effect.
    
    3. Usage is higher on weekdays for annual members, indicating that bike sharing is frequently used for commuting purposes.
    
    4. Casual users contribute relatively more trips during weekends, reflecting recreational use.
    
    5. The time-of-day analysis reveals pronounced peaks during morning and evening periods, further supporting the interpretation that annual memberships are linked to work-related travel.
    
    6. The station demand map identifies areas of concentrated bike activity, primarily in downtown Toronto, where population density and employment opportunities are highest.
    Overall, the visualization tells a story about how bike-sharing supports both urban mobility and recreational activities within Toronto.
    
    > What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots? 
    
    Several visualization design principles guided the development of this dashboard.
    
    First, appropriate chart selection was essential. A map was used to display station demand because geographic information is best interpreted spatially. Bar charts were selected to compare trip counts across categories such as month, day of the week, trip duration, and time of day because they facilitate straightforward comparisons.
    
    Second, visual hierarchy was considered. The station demand map occupies the largest area of the dashboard because it represents the central analytical component. Supporting charts are positioned around it to provide contextual information.
    
    Third, consistent colour encoding was applied throughout the dashboard. Blue represents annual members and orange represents casual members in every chart. This consistency reduces cognitive load and allows viewers to quickly identify patterns across different visualizations.
    
    Finally, the dashboard incorporates interactivity. Users can filter the visualizations by selecting membership categories, enabling exploration of subgroup differences without overwhelming viewers with excessive information.

    > How did you ensure that your data visualizations are reproducible? If the tool you used to make your data visualization is not reproducible, how will this impact your data visualization? 
    
    Reproducibility was ensured by conducting all stages of the analysis using Python scripts. The workflow included importing the raw dataset, data wrangling and transforming variables, and creating the final visualizations programmatically. The code was thoroughly commented and organized into logical sections to facilitate understanding and reuse.
    
    Additionally, the dataset source, software used, and required Python libraries were documented. This approach allows other researchers to reproduce the analysis or adapt the workflow to updated Bike Share Toronto datasets.
    
    If the visualization had been created exclusively using a point-and-click interface without preserving the underlying code, reproducibility would have been compromised. Future updates or modifications would require significant manual effort, increasing the likelihood of inconsistencies and errors.

    > How did you ensure that your data visualization is accessible?  
    
    Several measures were taken to improve accessibility.
    
    The visualization employs high-contrast colours that distinguish annual and casual members while remaining interpretable when viewed in grayscale. Labels and titles are clear and descriptive, reducing reliance on legends alone.
    Interactive elements allow users to focus on specific subsets of the data, minimizing visual clutter. Large font sizes and adequate spacing improve readability.
    
    However, accessibility limitations remain. Individuals with certain forms of colour vision deficiency may still encounter challenges differentiating categories. Future improvements could include the addition of texture patterns, direct data labels, alternative text descriptions, and compatibility with screen-reader technologies.

    > Who are the individuals and communities who might be impacted by your visualization?  
    
    The findings may influence several groups.
    
    Bike Share Toronto administrators could use the information to optimize bicycle redistribution, station placement, and membership strategies. Municipal governments may rely on these insights when making decisions about cycling infrastructure investments, such as expanding bike lanes or introducing new stations.
    
    Cyclists and local communities could also be affected. Areas identified as high-demand locations may receive enhanced services, whereas neighbourhoods with lower observed ridership might be overlooked despite potential unmet transportation needs. Therefore, decision-makers should interpret the findings alongside broader considerations of equity and access.

    > How did you choose which features of your chosen dataset to include or exclude from your visualization? 
    
    The dataset contained numerous variables related to bike-sharing activity. Features included in the dashboard were selected based on their relevance to the overall narrative of ridership behaviour derived from literature review.
    
    Variables such as station location, trip duration, month, day of the week, time of day, and membership type were included because they directly support questions about when, where, and by whom the bike-sharing system is used.
    
    Other variables were excluded because they contributed little to the intended message or risked overwhelming the audience. The goal was to balance analytical depth with clarity, ensuring that the dashboard remained interpretable and focused.

    > What ‘underwater labour’ contributed to your final data visualization product?
    
    A significant amount of "underwater labour" contributed to the final product. This invisible work included sourcing the Bike Share Toronto dataset, assessing data quality, cleaning missing or inconsistent values, standardizing date and time formats, deriving variables such as month and time categories, and testing different visualization approaches.
    
    Additional effort was required to refine the dashboard layout, select appropriate colour schemes, improve accessibility, and troubleshoot technical issues related to interactive functionality. Documentation and code annotation were also important components of the project, supporting reproducibility and future reuse.
    
    Although often overlooked, this preparatory work formed the foundation of the final visualization and significantly influenced its quality and interpretability.
